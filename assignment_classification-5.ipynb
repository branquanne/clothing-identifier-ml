### The $k$-NN model

You have to implement the $k$-NN classifier yourself. The below implementation uses the scikit-learn implementation of the $k$-NN classifier, but you need to do it yourself. Also, perform a hyper-parameter search for the optimal number of neighbors, $k$.

````python
# Create the k-NN classifier


# TODO: Create your own nearest neighbor classifier:
from collections import Counter
import numpy as np


class KNeighborsClassifier:

    def __init__(self, n_neighbors=5, batch_size=128):
        if n_neighbors < 1:
            raise ValueError("n_neighbours must be at least 1")

        self.n_neighbors = n_neighbors
        self.batch_size = batch_size # Using batching to speed up training
        self.X_train = None
        self.y_train = None

    def fit(self, X, y):
        X = np.asarray(X, dtype=np.float32)
        y = np.asarray(y)
        
        self.X_train = X
        self.y_train = y
        self.train_sq_norms = np.sum(X * X, axis=1)

        return self

    def predict(self, X):
        if self.X_train is None or self.y_train is None:
            raise ValueError("Model must be fitted before prediction")
        
        X = np.asarray(X, dtype=np.float32)
        
        predictions = []
        
        for start in range(0, len(X), self.batch_size):
            X_batch = X[start:start + self.batch_size]
            
            batch_sq_norms = np.sum(X_batch * X_batch, axis=1)
            distances = (batch_sq_norms[:, None] + self.train_sq_norms[None, :] - 2 * X_batch @ self.X_train.T)
            
            nearest_indecies = np.argpartition(distances, self.n_neighbors -1, axis=1)[:, :self.n_neighbors]
            nearest_labels = self.y_train[nearest_indecies]
            
            for labels in nearest_labels:    
                prediction = Counter(labels).most_common(1)[0][0]
                predictions.append(prediction)
            
        return np.asarray(predictions)

    def score(self, X, y):
        y = np.asanyarray(y)
        predictions = self.predict(X)
        
        return np.mean(predictions == y)
````

````python
model = KNeighborsClassifier(n_neighbors=5, batch_size=128)
model.fit(Xtrain[:50000], ytrain[:50000])
predictions = model.predict(Xval[:10])
accuracy = model.score(Xval[:1000], yval[:1000])

print("Predictions:\t ", predictions)
print("Actual labels:\t ", yval[:10])
print("Validation accuracy: ", accuracy)
````
